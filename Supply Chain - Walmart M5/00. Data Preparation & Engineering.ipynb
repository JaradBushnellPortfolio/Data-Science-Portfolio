This script demonstrates data sourcing and the transformations needed to support downstream modeling, such as demand forecasts and inventory optimization.

```python
import glob
from pathlib import Path

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import datetime as dt

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

project_dir = Path.cwd()
data_dir = project_dir / 'Raw Data'
```

```python
# load main datasets
# use a single store throughout this case study
sales = pd.read_csv(data_dir / 'sales_train_validation.csv', low_memory=False)






sales = sales[sales['store_id'] == 'CA_1'].reset_index().copy() # <-------------- use a single store for this example






calendar = pd.read_csv(data_dir / 'calendar.csv')
sell_prices = pd.read_csv(data_dir / 'sell_prices.csv')
```

```python
# convert sales from wide to long format
sales_long = sales.melt(
    id_vars=['id', 'item_id', 'dept_id', 'cat_id', 'store_id', 'state_id'],
    var_name='d',
    value_name='units_sold'
)

# add calendar dates and features
sales_long = sales_long.merge(
    calendar[
        ['d', 'date', 'wm_yr_wk', 'weekday', 'wday', 'month', 'year',
         'event_name_1', 'event_type_1']
    ],
    on='d',
    how='left',
    validate='many_to_one'
)

sales_long = sales_long[sales_long['d']!='index']
```

```python
sales_long['date'] = pd.to_datetime(sales_long['date'])

# add selling prices
sales_long = sales_long.merge(
    sell_prices[['store_id', 'item_id', 'wm_yr_wk', 'sell_price']],
    on=['store_id', 'item_id', 'wm_yr_wk'],
    how='left',
    validate='many_to_one'
)

sales_long['sell_price'] = sales_long['sell_price'].fillna(0)
```

```python
# check whether dates are consecutive
dates = sales_long['date'].drop_duplicates().sort_values()

expected_dates = pd.date_range(
    start=dates.min(),
    end=dates.max(),
    freq='D'
)

missing_dates = expected_dates.difference(dates)

if len(missing_dates) > 0:
    raise ValueError(f'missing dates: {missing_dates.tolist()}')

print('all dates are consecutive')
```

## sell_price

```python
# average selling price by date and category
grp = (
    sales_long.groupby(['d', 'cat_id'], as_index=False)
    ['sell_price'].mean()
)

# consistent colors by category
categories = sorted(grp['cat_id'].dropna().unique())
palette = plt.get_cmap('tab10')
colors = {
    category: palette(i % 10)
    for i, category in enumerate(categories)
}
```

```python
# plot price distributions by category
fig, ax = plt.subplots(figsize=(9, 5))

for category, data in grp.groupby('cat_id'):
    ax.hist(
        data['sell_price'].dropna(),
        bins=40,
        alpha=0.5,
        label=category,
        edgecolor='black',
        color=colors[category]
    )

ax.set(
    title='Distribution of Average Selling Prices by Category',
    xlabel='Average Selling Price',
    ylabel='Frequency'
)
ax.legend(title='Category')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print('We definitely need to investigate these three spikes.')
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=grp, x='cat_id', y='sell_price', palette='Set2')
plt.title('Average Selling Price by Category')
plt.xlabel('Category')
plt.ylabel('Average Selling Price')
plt.tight_layout()
plt.show()

print('FOODS is clearly it\'s own group, while HOBBIES has the widest range of avg selling prices.\
\nInteresting that HOBBIES and HOUSEHOLD have similiar means despite the difference in ranges.')
```

```python
# view the 90th–100th percentiles of selling price
percentiles = sales_long['sell_price'].describe(
    percentiles=np.arange(0, 1.01, 0.01)
).loc['90%':'100%']

print(percentiles)

percentiles.plot(
    kind='bar',
    figsize=(9, 4),
    title='Selling Price: 90th–100th Percentiles',
    edgecolor='black',
    grid=True
)

plt.xlabel('Percentile')
plt.ylabel('Selling Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print('Next step is understanding the outlier at $30.98, and possibly removing or capping it.')
```
